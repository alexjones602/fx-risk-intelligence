```python
# ==========================================
# AI-POWERED FX RISK INTELLIGENCE PLATFORM
# Week 3 - Session 1
# Feature Engineering
# ==========================================

import pandas as pd
import numpy as np

from pathlib import Path

# Load cleaned dataset
df = pd.read_csv("../data/processed/cleaned_fx_data.csv")

# Display first five observations
print(df.head())

# Display dataset dimensions
print("Dataset shape:")
print(df.shape)

# Display column names
print("\nColumn names:")
print(df.columns.tolist())

# Display data types
print("\nData types:")
print(df.dtypes)

# Display data range
print("\nFirst date:", df["Date"].iloc[0])
print("\nLast date:", df["Date"].iloc[-1])

# Display summary statistics
print("\nSummary statistics:")
print(df.describe())
```

```python
# Covert Date column to datetime
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

# Sort chronologically
df = df.sort_values("Date").reset_index(drop=True)

# Check date conversion
print(df["Date"].dtype)

# Check for invalid dates
print("Invalid dates:", df["Date"].isna().sum())
```

```python
# Ensure exchange rates are in numerical format
currency_pairs = ["GBPUSD","EURUSD","USDJPY"]
for pair in currency_pairs:
    column = f"{pair}_Close"

    df[column] = pd.to_numeric(df[column],errors="coerce")
```

```python
# Create daily percentage returns
for pair in currency_pairs:
    price_column = f"{pair}_Close"
    df[f"{pair}_Return_1D"] = (df[price_column].pct_change())

# Display daily percentage returns for GBPUSD
print(df[["Date","GBPUSD_Close","GBPUSD_Return_1D"]].head(10))

# Display daily percentage returns for EURUSD
print(df[["Date","EURUSD_Close","EURUSD_Return_1D"]].head(10))

# Display daily percentage returns for USDJPY
print(df[["Date","USDJPY_Close","USDJPY_Return_1D"]].head(10))
```

```python
# Calculate five-day percentage returns
for pair in currency_pairs:
    price_column = f"{pair}_Close"
    df[f"{pair}_Return_5D"] = (df[price_column].pct_change(periods=5))
```

```python
# Calculating rolling volatility
for pair in currency_pairs:
    returns_column = f"{pair}_Return_1D"

    # Five-day rolling volatility
    df[f"{pair}_Volatility_5D"] = (df[returns_column].rolling(window=5).std())

    # Twenty-day rolling volatility 
    df[f"{pair}_Volatility_20D"] = (df[returns_column].rolling(window=20).std())

    # Sixty-day rolling volatility 
    df[f"{pair}_Volatility_60D"] = (df[returns_column].rolling(window=60).std())
```

```python
# Calculating momentum
for pair in currency_pairs:
    price_column = f"{pair}_Close"

    # Five-day momentum
    df[f"{pair}_Momentum_5D"] = (df[price_column] / df[price_column].shift(5) - 1)
    
    # Twenty-day momentum
    df[f"{pair}_Momentum_20D"] = (df[price_column] / df[price_column].shift(20) - 1)
    
```

```python
# Display the newly created GBPUSD features
gbpusd_features = ["Date","GBPUSD_Return_1D","GBPUSD_Return_5D","GBPUSD_Volatility_5D","GBPUSD_Volatility_20D","GBPUSD_Volatility_60D","GBPUSD_Momentum_5D","GBPUSD_Momentum_20D"]
print(df[gbpusd_features])

# Display the newly created EURUSD features
eurusd_features = ["Date","EURUSD_Return_1D","EURUSD_Return_5D","EURUSD_Volatility_5D","EURUSD_Volatility_20D","EURUSD_Volatility_60D","EURUSD_Momentum_5D","EURUSD_Momentum_20D"]
print(df[eurusd_features])

# Display the newly created USDJPY features
jpyusd_features = ["Date","USDJPY_Return_1D","USDJPY_Return_5D","USDJPY_Volatility_5D","USDJPY_Volatility_20D","USDJPY_Volatility_60D","USDJPY_Momentum_5D","USDJPY_Momentum_20D"]
print(df[jpyusd_features])
```

```python
# Calculate targets
FORECAST_HORIZON = 5
for pair in currency_pairs:
    returns_column = f"{pair}_Return_1D"
    target_column = f"{pair}_Target_Volatility_5D"

    # Calculate forward-looking five-day volatility
    future_returns = df[returns_column].shift(-1)
    df[target_column] = (future_returns.iloc[::-1].rolling(window=FORECAST_HORIZON,min_periods=FORECAST_HORIZON).std().iloc[::-1])
```

```python
# Display GBPUSD target
print(df[["Date","GBPUSD_Return_1D","GBPUSD_Target_Volatility_5D"]].tail(15))

# Display EURUSD target
print(df[["Date","EURUSD_Return_1D","EURUSD_Target_Volatility_5D"]].tail(15))

# Display USDJPY target
print(df[["Date","USDJPY_Return_1D","USDJPY_Target_Volatility_5D"]].tail(15))
```

```python
# Count missing values for each column
missing_values = df.isnull().sum()
print("Missing values by column:")
print(missing_values[missing_values > 0])
```

```python
# Check duplicates
duplicate_dates = df["Date"].duplicated().sum()
print("Duplicated dates:", duplicate_dates)
print("Duplicated rows:", df.duplicated().sum())
```

```python
# Check invalid dates
print("Invalid dates:",df["Date"].isna().sum())

# Check chronological ordering
print("Dates in chronological order:",df["Date"].is_monotonic_increasing)

# Check unique dates
print("Unique dates:",df["Date"].nunique())
```

```python
# Check exchange rate validity
for pair in currency_pairs:
    column = f"{pair}_Close"
    invalid_prices = (df[column].isna() | (df[column] <= 0))
    print(pair, "Invalid pirces:",invalid_prices.sum())
```

```python
# Check volatility calculations
for pair in currency_pairs:
    volatility_columns = [f"{pair}_Volatility_5D",f"{pair}_Volatility_20D",f"{pair}_Volatility_60D",f"{pair}_Target_Volatility_5D"]
for column in volatility_columns:
    negative_values = (df[column] < 0).sum()
    print(column,"Negative values:",negative_values)
```

```python
# List features
feature_names = ["Return_1D","Return_5D","Volatility_5D","Volatility_20D","Volatility_60D","Momentum_5D","Momentum_20D"]
all_feature_columns = []
for pair in currency_pairs:
    for feature in feature_names:
        all_feature_columns.append(f"{pair}_{feature}")
print("Total predictive features:",len(all_feature_columns))
print(all_feature_columns)
```

```python
# Preparation for final machine-learning dataset
target_columns = [f"{pair}_Target_Volatility_5D" for pair in currency_pairs]
required_columns = (["Date"] + all_feature_columns + target_columns)

# Create machine-learning dataset
ml_df = df[required_columns].copy()

# Remove rows with missing features or targets
ml_df = ml_df.dropna(subset=all_feature_columns + target_columns)

# Reset index
ml_df = ml_df.reset_index(drop=True)
print("Original observations:",len(df))
print("ML observations:",len(ml_df))
print("Features:",len(all_feature_columns))
print("Targets:",len(target_columns))
print("\nFinal dataset shape:")
print(ml_df.shape)
```

```python
# Calculate initial split positions
n = len(ml_df)
train_end = int(n * 0.70)
validation_end = int(n * 0.85)

# Five-day purge period
gap = 5

# Training dataset
train_df = ml_df.iloc[:train_end].copy()

# Validation dataset
validation_df = ml_df.iloc[train_end + gap:validation_end].copy()

# Testing dataset
test_df = ml_df.iloc[validation_end + gap:].copy()

# Display dataset sizes
print("Training observations:",len(train_df))
print("Validation observations:", len(validation_df))
print("Testing observations:",len(test_df))
```

```python
print("Training period")
print(train_df["Date"].min(),"to",train_df["Date"].max())
print("\nValidation period")
print(validation_df["Date"].min(),"to",validation_df["Date"].max())
print("\nTesting period")
print(test_df["Date"].min(),"to",test_df["Date"].max())
```
